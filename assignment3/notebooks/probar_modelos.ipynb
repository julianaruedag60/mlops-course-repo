# Probar modelos de pingüinos desde MLflow
Ejecuta cada celda con **Shift + Enter**.

```python
import mlflow
import pandas as pd
```

```python
model = mlflow.pyfunc.load_model("models:/penguins-random-forest/1")
```

```python
pinguino = pd.DataFrame([{
    "island": "Biscoe",
    "bill_length_mm": 45.0,
    "bill_depth_mm": 15.0,
    "flipper_length_mm": 220.0,
    "body_mass_g": 5000.0,
    "sex": "MALE",
}])
model.predict(pinguino)
```

```python
# Comparar con el otro modelo
lr = mlflow.pyfunc.load_model("models:/penguins-logistic-regression/1")
lr.predict(pinguino)
```
